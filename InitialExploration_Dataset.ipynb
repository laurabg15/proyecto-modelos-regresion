# Introducción

### Problema Seleccionado

Queremos estimar el precio de venta de viviendas residenciales en **Ames, Iowa (EE. UU.)** a partir de sus características físicas, su ubicación y las condiciones de la venta. Este proyecto corresponde a la competencia de Kaggle [House Prices – Advanced Regression Techniques](https://www.kaggle.com/competitions/house-prices-advanced-regression-techniques/data).

### Objetivo del modelo
Construir un modelo que prediga el precio final de venta (`SalePrice`) de una vivienda. El modelo debe identificar qué variables influyen más en el valor y equivocarse lo menos posible con viviendas que no vio en el entrenamiento.

### Tipo de problema
Es un problema de **aprendizaje supervisado de regresión**, ya que la variable objetivo es continua y numérica.

### Variable objetivo
* **`SalePrice`:** Es el precio de venta de la vivienda, medido en dólares estadounidenses (USD).

### Fuente de los datos
Los datos provienen de la competencia de Kaggle [House Prices – Advanced Regression Techniques](https://www.kaggle.com/competitions/house-prices-advanced-regression-techniques/data). 

El conjunto original es el *Ames Housing Dataset*, recopilado por **Dean De Cock** (Truman State University, 2011) como una alternativa moderna al clásico dataset *Boston Housing*. Este registro recoge las ventas de viviendas ocurridas en Ames, Iowa, entre los años 2006 y 2010.


# Descripción del conjunto de datos

#### Número de observaciones
* **train.csv:** 1.460 viviendas (incluye la variable objetivo).
* **test.csv:** 1.459 viviendas (no incluye `SalePrice`, ya que Kaggle lo usa para evaluar).

#### Número de variables
* **train.csv:** Tiene **81 columnas**.
  * 1 identificador (`Id`).
  * 79 variables explicativas.
  * 1 variable objetivo (`SalePrice`).
* **test.csv:** Tiene **80 columnas** (excluye `SalePrice`).

#### Significado de la variable objetivo
* **SalePrice:** Es el precio de venta de cada vivienda (en USD).

#### Tipo de variables
El conjunto de datos contiene dos grandes grupos de variables:

* **43 variables categóricas** (tipo texto/object):
  * *Nominales:* Ejemplos como `Neighborhood`, `MSZoning`, `HouseStyle` y `SaleCondition`.
  * *Ordinales de calidad o condición:* Utilizan escalas estructuradas (Po < Fa < TA < Gd < Ex). Ejemplos: `ExterQual`, `KitchenQual`, `BsmtQual` y `HeatingQC`.

* **38 variables numéricas** (35 enteras y 3 decimales, incluyendo `Id` y `SalePrice`):
  * *Continuas:* Medidas de espacio como `GrLivArea`, `LotArea`, `TotalBsmtSF`, `GarageArea` y `LotFrontage`.
  * *Discretas:* Conteos como `BedroomAbvGr`, `FullBath`, `GarageCars` y `Fireplaces`.
  * *Temporales:* Fechas como `YearBuilt`, `YearRemodAdd`, `GarageYrBlt`, `MoSold` y `YrSold`.
  * *Numéricas con significado categórico:* Códigos o escalas como `MSSubClass` (tipo de vivienda codificado) y `OverallQual`/`OverallCond` (calificaciones del 1 al 10).

# Separación Entrenamiento y Prueba

Los datos de la competencia de Kaggle ya vienen divididos en un conjunto de entrenamiento y uno de prueba, este último sin la variable objetivo (SalePrice), por lo que funciona como un conjunto de validación externo para el modelo. Por esta razón, la exploración y preparación de datos se realiza únicamente sobre el conjunto de entrenamiento (tran.csv), el cual se divide de forma aleatoria en 80% para entrenamiento y 20% para prueba, con el fin de poder evaluar el desempeño del modelo antes de aplicarlo al conjunto de validación.

***Metodo utilizado:***
train_test_split (particion aleatoria) de scikit-learn


```python
import pandas as pd


train_test_df = pd.read_csv("train.csv")
validation_df = pd.read_csv("test.csv")

from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(
    train_test_df, test_size=0.2, random_state=80
)

print("Metodo utilizado: train_test_split (particion aleatoria) de scikit-learn")


print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)
print("Validation shape:", validation_df.shape)
print("Diferencia de columnas:", train_df.shape[1] - validation_df.shape[1], "(train tiene SalePrice, test no)")
print(f"Porcentaje para entrenamiento: {len(train_df) / len(train_test_df):.0%}")
print(f"Porcentaje para prueba: {len(test_df) / len(train_test_df):.0%}")
```

#### Eliminación de variables categóricas según la distribución de la variable objetivo

Para cada variable categórica comparamos la distribución de `SalePrice` entre sus categorías. Así pudimos ver qué variables permiten distinguir viviendas de precios distintos. Eliminamos las que no aportan información útil al modelo, teniendo en cuenta dos criterios:

1. **Variables casi constantes:** una sola categoría concentra casi todas las observaciones y las demás tienen muy pocos registros. En esta situación la variable prácticamente no varía y no puede explicar cambios en el precio. Además, al codificarlas se generan columnas casi vacías que añaden ruido y dimensionalidad. Es el caso de `Street`, `Utilities`, `LandContour`, `Condition1`, `Condition2`, `Heating` y `PavedDrive`.

2. **Variables sin diferencias en la variable objetivo:** las curvas de `SalePrice` de todas las categorías se superponen y tienen formas y centros similares. Es decir, pertenecer a una u otra categoría no cambia el precio de manera apreciable. Es el caso de `LotShape`, `LotConfig`, `LandSlope`, `BldgType`, `HouseStyle`, `RoofStyle`, `RoofMatl`, `Exterior1st`, `Exterior2nd`, `ExterCond`, `Foundation`, `BsmtCond`, `BsmtExposure`, `BsmtFinType1`, `BsmtFinType2`, `HeatingQC`, `Electrical`, `Functional`, `FireplaceQu`, `GarageType`, `GarageQual`, `GarageCond`, `Fence`, `SaleType` y `SaleCondition`.

Descartar estas variables reduce la complejidad y la dimensionalidad del modelo y disminuye el riesgo de sobreajuste. Con ello priorizamos las variables que sí muestran una relación clara con el precio de venta.

### Codificación de Variables Categoricas finales

Se lleva al one-hot-encoding las variables central air y GarageFinish


```python
import numpy as np

def to_onehot(x, values):
    r = np.r_[[np.argwhere(i==values)[0][0] for i in x]]
    return np.eye(len(values))[r].astype(int)


def replace_column_with_onehot(d, col, values):
    assert sum(d[col].isna())==0, "column must have no NaN values"
    k = to_onehot(d[col].values, values)
    r = pd.DataFrame(k, columns=["%s_%s"%(col, values[i]) for i in range(k.shape[1])], index=d.index).join(d)
    del(r[col])
    return r

garagefinish_categories = np.unique(train_df_cleaned["GarageFinish"])
centralair_categories = np.unique(train_df_cleaned["CentralAir"])

train_df_cleaned = replace_column_with_onehot(train_df_cleaned, "GarageFinish", garagefinish_categories)
train_df_cleaned = replace_column_with_onehot(train_df_cleaned, "CentralAir", centralair_categories)

train_df_cleaned.head(10)
```

# Modelo

### Modelo Base

```python
train_df_cleaned.head()
```

Se escoge un modelo de mediana, dado que la distribución de variable objetivo tiene valores extremos, lo cual la media como modelo base tendría peor desempeño

```python
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Separacion de features y target a partir de train_df_cleaned
X_train = train_df_cleaned.drop(columns=[ "SalePrice"])
y_train = train_df_cleaned["SalePrice"]

# Modelo base: siempre predice la mediana de SalePrice
baseline_model = DummyRegressor(strategy="median")
baseline_model.fit(X_train, y_train)

y_pred_baseline = baseline_model.predict(X_train)

print("Mediana predicha (constante):", round(float(baseline_model.constant_[0][0]), 2))
print("MAE  baseline:", round(mean_absolute_error(y_train, y_pred_baseline), 2))
print("RMSE baseline:", round(np.sqrt(mean_squared_error(y_train, y_pred_baseline)), 2))
print("R2   baseline:", round(r2_score(y_train, y_pred_baseline), 2))
```

### Modelo Entrenamiento y Validación Cruzada

Para elegir el modelo y sus hiperparámetros usamos **validación cruzada de 5 folds dentro del conjunto de entrenamiento**. El conjunto de prueba (`test_df`) no se usa en ninguna decisión: se consulta **una sola vez**, al final, para evaluar el modelo elegido.

Para evitar fuga de información, todas las transformaciones viven en la clase `TransformadorCasas` (archivo `transformador_casas.py`) y se combinan con cada modelo en un `Pipeline`. Así, en cada fold la imputación, la codificación one-hot y el escalado se ajustan **solo con los folds de entrenamiento** y luego se aplican al fold de validación.

Limitaciones que reconocemos:
* La selección de variables por correlación se hizo una vez sobre todo el entrenamiento. En la celda siguiente verificamos que, recalculada dentro de cada fold, selecciona exactamente las mismas variables.
* El descarte de variables categóricas se basó en la inspección visual del EDA sobre el entrenamiento y no puede repetirse automáticamente por fold.
* El atípico (Id 1299) se retira del entrenamiento antes de la validación cruzada.

```python
from sklearn.model_selection import KFold

kfold = KFold(n_splits=5, shuffle=True, random_state=80)


def seleccion_por_correlacion(df, umbral=0.6):
    corr = df.select_dtypes(include="number").corr()["SalePrice"].drop(["SalePrice", "Id"])
    return sorted(corr[corr.abs() >= umbral].index)


seleccion_total = seleccion_por_correlacion(train_df_sin_atipico)
print("Seleccion con todo el entrenamiento:", seleccion_total)
for i, (idx_entrenamiento, _) in enumerate(kfold.split(train_df_sin_atipico)):
    seleccion_fold = seleccion_por_correlacion(train_df_sin_atipico.iloc[idx_entrenamiento])
    print(f"Fold {i}: igual a la seleccion total = {seleccion_fold == seleccion_total}")
```

```python
from sklearn.model_selection import cross_validate
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor
from sklearn.dummy import DummyRegressor
from xgboost import XGBRegressor
import pandas as pd

from transformador_casas import TransformadorCasas

X_cv = train_df_sin_atipico.drop(columns=["SalePrice"])
y_cv = train_df_sin_atipico["SalePrice"]

metricas = {
    "MAE": "neg_mean_absolute_error",
    "RMSE": "neg_root_mean_squared_error",
    "R2": "r2",
}


def evaluar_con_cv(modelos):
    filas = []
    for nombre, modelo in modelos.items():
        pipeline = Pipeline([("transformador", TransformadorCasas()), ("modelo", modelo)])
        r = cross_validate(pipeline, X_cv, y_cv, cv=kfold, scoring=metricas, return_train_score=True)
        signo = {"MAE": -1, "RMSE": -1, "R2": 1}
        fila = {"Modelo": nombre}
        for m, s in signo.items():
            fila[f"{m} Train"] = s * r[f"train_{m}"].mean()
            fila[f"{m} Val"] = s * r[f"test_{m}"].mean()
        fila["MAE Val (desv.)"] = r["test_MAE"].std()
        filas.append(fila)
    return pd.DataFrame(filas).round(2)


modelos = {
    "Modelo Base (Mediana)": DummyRegressor(strategy="median"),
    "Regresion Lineal": LinearRegression(),
    "Ridge": Ridge(),
    "Lasso": Lasso(),
    "Random Forest Regressor": RandomForestRegressor(random_state=80),
    "XGBoost Regressor": XGBRegressor(random_state=80),
}

comparacion_cv_df = evaluar_con_cv(modelos)
comparacion_cv_df
```

**Análisis de la primera comparación (validación cruzada).** Todos los modelos mejoran ampliamente al modelo base (MAE de validación ≈ USD 56.000). Al comparar el desempeño en entrenamiento y en validación se distinguen dos comportamientos:

- **Random Forest y XGBoost muestran sobreajuste:** su error en entrenamiento es mucho menor que en validación (MAE ≈ USD 7.700 → 20.700 en Random Forest y ≈ USD 2.300 → 22.000 en XGBoost; R² de 0,98 y 1,00 en entrenamiento frente a 0,81 y 0,79 en validación). Es decir, memorizan buena parte del conjunto de entrenamiento, y su desempeño en los datos que ven no es un buen indicador de su desempeño en datos nuevos.
- **Los modelos lineales son estables:** la diferencia entre entrenamiento y validación es pequeña (MAE ≈ USD 23.500 → 23.900; R² 0,81 → 0,79), lo que indica que no memorizan los datos y que su comportamiento es predecible.

Reconocemos que, en validación cruzada, Random Forest obtiene un MAE menor (≈ USD 20.700 frente a ≈ USD 23.900) y un RMSE menor. Una brecha amplia entre entrenamiento y validación no demuestra por sí sola que un modelo vaya a desempeñarse peor en el futuro; sin embargo, sí indica mayor varianza y mayor dependencia de las viviendas particulares con que se entrena.

A partir de esto, pasamos a una segunda competencia con los modelos lineales regularizados: probamos Ridge y Lasso con distintos valores de alpha (0.1, 1, 10 y 100), usando como referencia el modelo base (mediana) y la Regresión Lineal sin regularización. Los valores de alpha también se comparan con validación cruzada, no con el conjunto de prueba.

```python
modelos_parametros = {
    "Modelo Base (Mediana)": DummyRegressor(strategy="median"),
    "Regresion Lineal": LinearRegression(),
    **{f"Ridge (alpha={a})": Ridge(alpha=a) for a in [0.1, 1, 10, 100]},
    **{f"Lasso (alpha={a})": Lasso(alpha=a, max_iter=10000) for a in [0.1, 1, 10, 100]},
}

comparacion_parametros_df = evaluar_con_cv(modelos_parametros)
comparacion_parametros_df
```

### Interpretación de Resultados

**Selección del modelo final: Lasso (alpha = 100).**

En la segunda competencia, la Regresión Lineal, Ridge y Lasso obtienen resultados prácticamente iguales para todos los valores de alpha: las diferencias de MAE en validación (≈ USD 300 como máximo) son mucho menores que la variación entre folds (desviación ≈ USD 1.500), por lo que ninguna configuración es claramente superior en error. Elegimos **Lasso con alpha = 100** por las siguientes razones:

1. **Menor riesgo de sobreajuste.** Lasso mantiene la brecha mínima entre entrenamiento y validación (MAE ≈ USD 23.500 → 23.900; R² 0,81 → 0,79), a diferencia de Random Forest y XGBoost. Además, la penalización L1 limita el tamaño de los coeficientes, lo que reduce la varianza del modelo frente a cambios en los datos de entrenamiento.
2. **Elimina variables redundantes.** Las variables one-hot de `GarageFinish` y `CentralAir` son colineales (por ejemplo, `CentralAir_N` + `CentralAir_Y` = 1). En la Regresión Lineal sin regularizar, el efecto se reparte de forma arbitraria entre ellas; Lasso lleva a cero los coeficientes redundantes (`CentralAir_Y` y `GarageFinish_Fin`), dejando un modelo más simple y con coeficientes interpretables.
3. **Interpretabilidad.** Cada coeficiente se lee directamente como el cambio en dólares del precio estimado, lo que permite explicar las predicciones.
4. **Mejor alpha dentro de Lasso.** Entre los valores probados, alpha = 100 obtiene el menor MAE (≈ USD 23.920) y el menor RMSE (≈ USD 36.313) de validación de la familia Lasso.

**Intercambio aceptado:** Random Forest obtiene un MAE de validación unos USD 3.200 menor. Elegimos conscientemente Lasso, priorizando estabilidad, simplicidad e interpretabilidad sobre esa diferencia de error.

**Interpretación de las métricas (validación cruzada):** en promedio, el modelo se equivoca alrededor de **USD 23.900** por vivienda (MAE), frente a ≈ USD 56.000 del modelo base. El RMSE de validación es ≈ **USD 36.300**; es mayor que el MAE porque penaliza más los errores grandes, que se concentran en las viviendas más caras. El R² de validación de ≈ **0,79** indica que el modelo explica alrededor del 79 % de la variabilidad del precio.

Para futuras versiones, se podría aplicar un logaritmo a `SalePrice` para reducir el efecto de los precios extremos.

### Evaluación Final en el Conjunto de Prueba

El modelo elegido se entrena con todo el conjunto de entrenamiento (sin el atípico) y se evalúa **una única vez** en el conjunto de prueba. Este mismo pipeline es el que se guarda en `modelo_casas.joblib`, por lo que las métricas reportadas corresponden exactamente al artefacto entregado.

```python
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

modelo_final = Pipeline([
    ("transformador", TransformadorCasas()),
    ("modelo", Lasso(alpha=100, max_iter=10000)),
])
modelo_final.fit(X_cv, y_cv)

X_test = test_df.drop(columns=["SalePrice"])
y_test = test_df["SalePrice"]
pred_test = modelo_final.predict(X_test)

print("MAE  test:", round(mean_absolute_error(y_test, pred_test), 2))
print("RMSE test:", round(np.sqrt(mean_squared_error(y_test, pred_test)), 2))
print("R2   test:", round(r2_score(y_test, pred_test), 3))

coeficientes = pd.Series(
    modelo_final.named_steps["modelo"].coef_,
    index=modelo_final.named_steps["transformador"].columnas_finales_,
).round(2)
print("\nCoeficientes del modelo (USD):")
print(coeficientes.sort_values(ascending=False))
```

En el conjunto de prueba, consultado una sola vez, el modelo obtiene un **MAE de ≈ USD 21.000**, un **RMSE de ≈ USD 27.700** y un **R² de ≈ 0,86**. Son resultados del mismo orden que los de la validación cruzada (algo mejores, dentro de la variación observada entre folds), lo que es coherente con la baja brecha entre entrenamiento y validación del modelo.
